```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python

df = pd.read_csv('data_core.csv')


numerical_features = ['Temparature', 'Humidity', 'Moisture', 'Nitrogen', 'Potassium', 'Phosphorous']
df_numeric = df[numerical_features]

df_numeric.describe()
```

```python

plt.figure(figsize=(10, 5))
sns.boxplot(data=df_numeric)
plt.title('Feature Distributions Before Scaling')
plt.ylabel('Original Units / Values')
plt.show()
```

```python

scaler = StandardScaler()
scaled_array = scaler.fit_transform(df_numeric)


df_scaled = pd.DataFrame(scaled_array, columns=numerical_features)


print("--- Original Data (First 5 Rows) ---")
print(df_numeric.head())

print("\n--- Scaled Data (First 5 Rows) ---")
print(df_scaled.head())
```
