```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/data_core.csv')

numerical_features = ["Temparature", "Humidity", "Moisture", "Nitrogen", "Potassium", "Phosphorous"]
df_numeric = df[numerical_features]

df_numeric.head()
```

```python
correlation_matrix = df_numeric.corr(method="pearson")

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap of Numerical Features")
plt.show()

correlation_matrix
```

```python
# Get correlation pairs, excluding self-correlation (diagonal)
corr_pairs = correlation_matrix.unstack()
corr_pairs = corr_pairs[corr_pairs != 1.0]

strongest_positive = corr_pairs.idxmax(), corr_pairs.max()
strongest_negative = corr_pairs.idxmin(), corr_pairs.min()

print("Strongest positive correlation:", strongest_positive)
print("Strongest negative correlation:", strongest_negative)
```
